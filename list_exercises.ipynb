```python
tools = ['laptop', 'monitor', 'keyboard','mouse']
```

# references
- creating a copy of a list does not create a new list. It points to the same place in memory
- Google: Pass by value vs Pass vs reference
- Changing one referenced list will also change the other
- This is important bc we need to keep the format list intact without changes, so we need to change the copied version 

```python
og_list = [1,2,3,4,5]

new_list = og_list 

print('OG', og_list)
print('New', new_list)

new_list.append(6)
print('OG', og_list)
print('New', new_list)  # this changed BOTH the new and og list because they are pointing to the same memory location (new_list = og_list)

```

```python
# Deep copy - a completely seperate copy of Non primative data types (lists, dictionaries, sets, etc) not just a refernece to the place in memory
# How to create a python list method to make a deep copy - copy.deepcopy 
second_copy = og_list.copy()
print(second_copy)
print(og_list)

og_list.append(7)
print(second_copy)
print(og_list)
```

```python
# reverse extends, short

num_list = [1,2,3,4,5]
print(num_list)

num_list.reverse()
print(num_list)

num_list.extend([6,7,8,9])
print(num_list)

num_list.sort()
print(num_list)
```

```python
# Create a function, take in 2 lists of years, reverse so it is in chronological order, add them together
# functions allow us to create reusable code blocks (print), (length), etc.
# def allows us to define a function 

def reorder_profit_years(years1, years2):
    # reverse the list to chronological order
    years1.reverse()
    years2.reverse()
    
    # add these two lists together
    years1.extend(years2)

    # pass a value back to the main program
    return years1
# We defined a function called reorder_profit_years that takes in 2 lists of years, 
# reverses the order of years, adds them together. 
# The function returns the combined list in chronological order.    
```

```python
# Now using the function we defined above

year_list1 = [2026, 2025, 2024, 2023]
year_list2 = [1990, 1989, 1987, 1986]

print(reorder_profit_years(year_list2, year_list1))
```
