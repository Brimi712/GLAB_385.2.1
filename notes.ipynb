# List notes and practice
### Steps to create Env and ipynb (Python Notebook)
1. Create a project folder.
2. Open project folder in VSC
3. Verify your explorer tab and terminal are in the right place
*** No nested repositories or projects ***
4. (Optional) initialize repo if pushing to github
5. Create a virtual environment
   - In terminal: `python -m venv .venv`
*** If it fails, delete and run again ***
6. Create '.gitignore' file at the ROOT of the project (next to the .venv, NOT in)
7. Document '.venv/' inside 'gitignore'
8. Activate environment 
    - In terminal: 'source .venv/Scripts/activate' 
9. Install ipykernel
    - In terminal: 'pip install ipykernel'
10. Create '.ipynb' file
11. Select Kernel (top right of ipnyb file once open)
    - Select the env you created (Should be recommended with a star)

#### Data Collections 
- Arrays (lsits) - "Indexed collections" - meaning they have order
- Dict (dictionaries/Objects) - "keyed collection" - No order, used to access information.  
- Tuples (records/rows in a table)


### String

- Collection of single characters

```python
greet = 'Hello World!'
#print the first value in a string/list
print(greet[0])
print(greet[1])
print(greet[2])
```

### Python indexing operator
- can be used with strings and lists
- by default if we do not include the step it is 1
[ start : stop : step ] 

```python
print(greet[0:5])
```

### Types of Error:
- 'Indexerror' - out of range, trying to access something that is not there. 
- 'Typeerror' - wrong type in a list index, wrong type in an operation. 
- 'syntaxerror' - incorrect code grammar
- 'namerror' - variable/function not defined
- valueerror' - correct type but incorrect value 
- 'keyerror' - dictionary key that does not exist

```python
people[100] # index error
people[1.5] # type error
if x > 5 print(x) # syntax error
if x > 5 print(x) # name error
int('hello') # value error
me ={}
me['name'] # key error
```
